```python
# Original median
median_before = df['horsepower'].median()
print("Median before filling:", median_before)
```

```python
# Most frequent value (mode)
most_frequent = df['horsepower'].mode()[0]
print("Most frequent value:", most_frequent)
```

```python
# Fill missing values with the mode
df['horsepower'] = df['horsepower'].fillna(most_frequent)
```

```python
# New median
median_after = df['horsepower'].median()
print("Median after filling:", median_after)
```

```python
# Select cars from Asia
asia_cars = df[df['origin'] == 'Asia']
```

```python
# Select the required columns and first 7 rows
X = asia_cars[['vehicle_weight', 'model_year']].head(7).to_numpy()
```

```python
# Create the y array
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
# Compute X transpose multiplied by X
XTX = X.T @ X
```

```python
# Invert XTX
XTX_inv = np.linalg.inv(XTX)
```

```python
# Calculate the regression coefficients
w = XTX_inv @ X.T @ y
```

```python
# Sum of all elements
w.sum()
```
